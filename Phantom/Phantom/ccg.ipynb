## Part I — The Abstract Framework

### 1.1  Setup: Starting from M

Let $H$ be a separable Hilbert space. Fix a self-adjoint operator $L$ on $H$ — the reference generator — providing a preferred spectral decomposition via its spectral projections $E_L(\Delta)$. In PDE applications $L$ is a differential operator (Laplacian, biharmonic, Helmholtz) and its eigenspaces are the modes of the system.

Fix a bounded measurement operator $M : H \to H$. No self-adjointness or positivity is assumed at this level. The observed state is:

$$y = Mf.$$

$M$ encodes the instrument's action: what it measures, what it misses, how it attenuates or transforms the true field $f$. The observation defect is:

$$Q := I - M,$$

so $M = I - Q$. The defect $Q$ does not need to be self-adjoint, positive, or a projection at this level of abstraction.

*Correction from v1: v1 opened with $Q$ as the primitive and assumed $Q = Q^*$, $0 \leq Q \leq I$ throughout Part I. This is the self-adjoint attenuation subclass, not the full framework. Opening with $M$ corrects this. The self-adjoint case is introduced explicitly in §1.4.*

---

### 1.2  The Two Fundamental Block Decompositions

Fix an orthogonal projection $P$ in $H$. In applications $P$ will be an eigenspace projector $P_\lambda$, a spectral cutoff $P_N$, or a resolution projector $R$. Relative to $P$, the measurement operator $M$ decomposes into two blocks for any $f \in PH$:

**Compressed (forward) block:**

$$A_P(M) := PMP$$

**Off-diagonal (fabrication) block:**

$$H_P(M) := (I-P)MP$$

For $f \in PH$:

$$Mf = A_P(M)f + H_P(M)f,$$

where $A_P(M)f \in PH$ and $H_P(M)f \in (I-P)H$. The observation splits exactly into content that stays in the model subspace and content that leaks out of it.

---

### 1.3  The Exact Fabrication Law

The Gram operator of the masked dictionary $\{M\varphi_n : \varphi_n \in PH\}$ is:

$$G_P(M) := PM^*MP.$$

Expanding:

$$G_P(M) = A_P(M)^*A_P(M) + H_P(M)^*H_P(M).$$

*Proof:* $PM^*MP = (PMP)^*(PMP) + ((I-P)MP)^*(\cdot)$ \ldots expanding $PM^*MP = PM^*(P + (I-P))MP = (PMP)^*(PMP) + (MP - PMP)^*(MP - PMP)$ restricted to $PH$, which gives $A_P^*A_P + H_P^*H_P$. $\square$

Define the fabrication covariance operator:

**Fabrication covariance (exact identity):**

$$\Sigma_P(M) := H_P(M)^*H_P(M) = G_P(M) - A_P(M)^*A_P(M).$$

For $f \in PH$, the total off-subspace content generated by the measurement is:

$$\|H_P(M)f\|^2 = \langle \Sigma_P(M) f, f \rangle.$$

This is exact for any bounded $M$ and any projection $P$. Fabrication energy is the deficit between the Gram operator of the masked dictionary and the squared forward map. It measures the failure of $M$ to preserve the geometry of $PH$.

---

### 1.4  Self-Adjoint Attenuation Subclass: $M = I - Q$

Restrict now to $M = I - Q$ with $Q = Q^*$, $0 \leq Q \leq I$ (a positive contraction). This is the self-adjoint attenuation class: instruments that attenuate but do not phase-shift. Real spatial masks, smooth apodization windows, and partial-visibility operators all belong here.

Under this assumption:

$$A_P(M) = P(I-Q)P = I_P - PQP =: I_P - T_P(Q),$$

$$H_P(M) = (I-P)(I-Q)P = -(I-P)QP =: -H_P(Q),$$

$$G_P(M) = P(I-Q)^2 P = I_P - 2T_P(Q) + PQ^2P.$$

And the fabrication covariance becomes:

$$\Sigma_P(Q) = PQ^2P - (PQP)^2 = H_P(Q)^*H_P(Q).$$

The fabrication law in terms of $Q$:

**Fabrication = compression variance (self-adjoint attenuation case):**

$$\|H_P(Q)f\|^2 = \langle PQ^2P \, f, f \rangle - \langle PQP \, f, f \rangle^2 \quad [\text{in 1D}] = \operatorname{Var}_f(Q \text{ restricted to } PH).$$

Fabrication energy is the variance of the defect observable $Q$ in the state $f$ relative to the subspace $PH$. Not a missing-mass law. A variance law.

Forward/Gram identity:

$$G_P = A_P^2 + \Sigma_P(Q).$$

The correct condition for $G_P = A_P$ is obtained from $G_P - A_P = PQ^2P - PQP = P(Q^2 - Q)P = -PQ(I-Q)P$:

**Forward/Gram coincidence condition (corrected v2→v3):**

$$G_P = A_P \iff PQ(I-Q)P = 0.$$

This is not the same as $\Sigma_P(Q) = 0$. Counterexample: $Q = \alpha I$ on $PH$ for $0 < \alpha < 1$. Then $\Sigma_P(Q) = \alpha^2 I - \alpha^2 I = 0$, but $PQ(I-Q)P = \alpha(1-\alpha)I \neq 0$, so $G_P = (1-\alpha)^2 I \neq (1-\alpha)I = A_P$. The condition $\Sigma_P(Q) = 0$ is strictly stronger than needed and is the wrong criterion for forward/Gram coincidence.

*Bug in v2: §1.4 stated "$G_P = A_P$ iff $\Sigma_P(Q) = 0$." Corrected in v3 to $PQ(I-Q)P = 0$, confirmed by ChatGPT analysis. The hard hole is unaffected: $Q = \chi_G$ gives $Q(I-Q) = Q - Q^2 = 0$, so the condition holds. The error was in the general attenuation claim.*

---

### 1.5  The Hard Hole: $Q = \chi_G$

The further specialization to $Q = \chi_G$ (indicator function of a spatial region $G$) adds idempotency: $Q^2 = Q$. Under this:

$$\Sigma_P(Q) = PQP - (PQP)^2 = T_P(Q)(I_P - T_P(Q)).$$

In one dimension with $T_P(Q) = \beta$ (scalar), this is $\beta(1-\beta)$ — Bernoulli variance. The maximum at $\beta = 1/2$ confirms: complete occlusion ($\beta = 1$) generates zero fabrication because no measured content exists to mislabel. Partial occlusion is worse than complete occlusion for phantom fabrication.

Idempotency also gives $G_P = A_P$ exactly (since $\Sigma_P(\chi_G) = T_P(I - T_P)$ and $A_P = I - T_P$, so $A_P^2 + \Sigma_P = (I-T_P)^2 + T_P(I-T_P) = (I-T_P)(I-T_P+T_P) = I-T_P = A_P$). The forward map and measurement geometry coincide for hard masks.

The cascade of rigidities in the hole case — Bernoulli variance, $G_P = A_P$, coincidence of exposure and fabrication bases — are all consequences of $Q^2 = Q$. They are not properties of observation defects in general. The hole is an unusually tractable corner of the full theory.

---

### 1.6  Fabrication as Noncommutativity

For the self-adjoint attenuation class, since $[P, Q]P = PQP - QP = -(I-P)QP = -H_P(Q)$:

**Fabrication = compressed commutator norm (exact):**

$$\Sigma_P(Q) = P[P,Q]^*[P,Q]P.$$

Zero fabrication criterion: $\Sigma_P(Q) = 0 \iff [P,Q]P = 0 \iff Q(PH) \subseteq PH$. No false off-subspace content is generated when the defect preserves the model subspace. For $Q = \chi_G$ and $P = P_\lambda$: this requires every eigenfunction in $E_\lambda$ to vanish inside $G$ — impossible generically.

*The statement in v1 that "$Q$ is diagonal in physical space and non-diagonal in spectral coordinates" belongs only to the multiplication-defect realization, where $Q = q(x)$ is a function of position and the eigenfunctions of $L$ provide a genuinely different coordinate system. At the abstract Hilbert space level, there is no distinguished physical space, and the statement is meaningless. It is moved to Part III where the multiplication-defect realization is developed explicitly.*

---

### 1.7  Basis Bifurcation for Non-Idempotent Defects

For $Q = \chi_G$ (idempotent), the fabrication operator $\Sigma_P(Q) = T_P(Q)(I - T_P(Q))$ is a polynomial in $T_P(Q)$. Therefore $T_P(Q)$ and $\Sigma_P(Q)$ share eigenvectors: the exposure basis and fabrication basis coincide.

For general positive contractions $Q$ with $Q^2 \neq Q$, the operators $T_P(Q) = PQP$ and $\Sigma_P(Q) = PQ^2P - (PQP)^2$ are not generally functions of each other and need not share eigenvectors. There are then two distinct bases inside $PH$:

- **Exposure basis:** eigenvectors of $T_P(Q) = PQP$. Sorts directions in $PH$ by how much the defect acts on them. Defined for the self-adjoint attenuation subclass.

- **Fabrication basis:** eigenvectors of $\Sigma_P(Q) = PQ^2P - (PQP)^2$. Sorts directions by how much false off-subspace content they generate. Coincides with exposure basis for hard holes ($Q^2 = Q$); generically distinct for soft defects.

- **Ambiguity basis — general $M$:** small-eigenvalue eigenspace of $G_P(M) = PM^*MP$. Sorts directions in $PH$ by how poorly $M|_{PH}$ can be inverted — i.e., which directions in the model class are nearly erased by the measurement. This is the correct recoverability geometry for arbitrary $M$.

- **Ambiguity basis — self-adjoint attenuation ($M = I-Q$):** $G_P = A_P^2 + \Sigma_P(Q)$. For the hard hole, $G_P = A_P$ (since $Q(I-Q) = 0$), and small eigenvalues of $G_P$ correspond to large eigenvalues of $C_N = PQP$. For soft defects with $Q(I-Q) \neq 0$, the ambiguity basis must be computed from $G_P$ directly, not from $T_P(Q)$ alone.

The three bases coincide for the hard hole (all reduce to the $C_N$ eigenbasis) and generically separate for soft defects. The clean factoring: general $M$ uses singular geometry of $M|_{PH}$; positive $Q$ lets you sharpen that into eigengeometry; idempotent $Q$ further collapses exposure, fabrication, and ambiguity into one basis.

*Correction from v2 (flagged by ChatGPT): v2 defined the ambiguity basis as "near-1 eigenspace of $T_{P_N}(Q)$" universally. This is only correct for the hard hole. For general $M$, recoverability is governed by the singular values of $M|_{PH}$, i.e., the eigenvalues of $G_P(M) = PM^*MP$ — not by $T_P(Q)$ except in the hard-hole subcase where $G_P = A_P$ collapses them. The distinction matters for soft defects and for non-self-adjoint $M$.*

### 1.8  The Abstract Decomposition Theorem

The hole decomposition $w_{G,N} = w_G + s_{G,N}$ is an instance of a more general law that holds for any bounded $M$, any model projector $P$, and any resolution projector $R$ with $[P,R] = 0$.

For $f \in PH \cap RH$, define:

$$w_P(f) := (I-P)Mf \qquad \text{(exact model-space lie)},$$

$$t_{P,R}(f) := (I-P)[R,M]f \qquad \text{(resolution-commutator correction)}.$$

**Abstract decomposition (exact, for $[P,R] = 0$, $f \in PH \cap RH$):**

$$(I-P)RMf = w_P(f) + t_{P,R}(f).$$

*Proof:* $(I-P)[R,M]f = (I-P)RMf - (I-P)MRf = (I-P)RMf - (I-P)Mf$ (using $Rf = f$ since $f \in RH$). So $t_{P,R} = (I-P)RMf - w_P$. Rearranging: $(I-P)RMf = w_P + t_{P,R}$. $\square$

The left side $(I-P)RMf$ is the off-model content of the resolution-limited observation of $Mf$. It splits exactly into two pieces:

- $w_P(f) = (I-P)Mf$: the exact global model-space lie. Contains all the off-model content in the full (unresolved) observation. Independent of $R$.

- $t_{P,R}(f) = (I-P)[R,M]f$: the resolution-commutator correction. Sourced entirely by the failure of $R$ to commute with $M$. Vanishes when $R = I$ (no bandwidth restriction). Vanishes when $[R,M] = 0$ ($M$ preserves the resolution structure).

The hole decomposition is the instance $P = P_\lambda$, $R = P_N$, $M = M_G$:

- $w_P = (I-P_\lambda)M_G f = w_G$ (the exact global modal lie).

- $t_{P,R} = (I-P_\lambda)[P_N, M_G]f = (I-P_\lambda)s_{G,N}$.

- The hole paper writes $w_{G,N} = w_G + s_{G,N}$ using the full commutator $s_{G,N} = [P_N,M_G]f$, not $(I-P_\lambda)s_{G,N}$. The difference: $P_\lambda s_{G,N}$ contributes to the intra-eigenspace part of the observation, not to the phantom. Both formulations are correct; the abstract version lives entirely in $(I-P)H$, the hole paper's version is operationally direct.

*Addition in v3 (proposed by ChatGPT analysis): the abstract decomposition lifts the hole's best concrete insight into the general $M$ framework. The hole decomposition previously appeared as a specific computation. The abstract version shows it is a universal law: exact model-space lie + resolution-commutator correction, for any measurement operator $M$ and any pair of commuting projectors.*

## Part II — Spectral Routing

### 2.1  Transfer Blocks and the Commutator Routing Law

For eigenspace projectors $P_\lambda$, $P_\mu$ of $L$ with $\lambda \neq \mu$, the transfer block is:

$$Q_{\mu\lambda} := P_\mu Q P_\lambda.$$

Using $L P_\lambda = \lambda P_\lambda$:

$$P_\mu [L, Q] P_\lambda = (\mu - \lambda) Q_{\mu\lambda}.$$

Iterating $k$ times:

**Iterated commutator routing law:**

$$(\mu - \lambda)^k \, Q_{\mu\lambda} = P_\mu \, \mathrm{ad}_L^k(Q) \, P_\lambda,$$
$$\mathrm{ad}_L^k(Q) = [L, [L, \ldots [L, Q] \ldots]] \quad (k \text{ nested commutators}).$$

If $\mathrm{ad}_L^k(Q)$ is bounded in operator norm:

$$\|Q_{\mu\lambda}\| \leq \|\mathrm{ad}_L^k(Q)\| / |\mu - \lambda|^k.$$

Spectral off-diagonal decay is controlled by the $L$-regularity of the defect. High $L$-regularity (many bounded commutators) gives fast spectral decay. Low regularity gives slow decay.

*Domain qualification (correction from v1): for unbounded $L$, the identity $(\mu-\lambda)^k Q_{\mu\lambda} = P_\mu \mathrm{ad}_L^k(Q) P_\lambda$ requires $\mathrm{ad}_L^k(Q)$ to be defined on the appropriate domain. In PDE applications with $L = -\Delta$ and $Q = \chi_G$, the first commutator $[-\Delta, \chi_G]$ is a boundary distribution on $\partial G$ — well-defined distributionally but not as a bounded operator on all of $H$. The identity holds in the distributional sense or on the dense domain where $\mathrm{ad}_L^k(Q)$ acts. The abstract section should not be read as implying these commutators are always bounded operators; in the multiplication-defect PDE setting they are boundary distributions.*

---

### 2.2  The Boundary Wronskian Formula

For $L = -\Delta$ on a bounded domain $\Omega$ and $Q = \chi_G$, Green's second identity applied to $\varphi_n$ and $\bar{\varphi}_m$ on $G$ gives the exact boundary Wronskian identity:

**Boundary Wronskian (exact, multiplication-defect realization):**

$$(\lambda_m - \lambda_n) K_{mn} = \int_{\partial G} [\bar{\varphi}_m \partial_\nu \varphi_n - \varphi_n \partial_\nu \bar{\varphi}_m] \, dS,$$

where $K_{mn} = \int_G \varphi_n \bar{\varphi}_m$. This is the $k=1$ case of the routing law made explicit via Green's identity. Off-diagonal coupling is exactly a boundary Wronskian divided by eigenvalue gap. Near-resonant modes (small $|\lambda_m - \lambda_n|$) receive larger coupling for the same boundary interaction. This formula belongs to the multiplication-defect PDE branch, not to the abstract framework.

## Part III — The Multiplication-Defect Realization

Parts I and II are abstract: $H$ is a general Hilbert space, $M$ a bounded operator, $Q$ a positive contraction. The following results require the additional structure of a multiplication defect: $Q = q(x)$ acting as pointwise multiplication in a function space, with $L$ a local differential operator. This structure provides physical space, eigenfunctions as functions on a domain, and the boundary geometry $\partial G$.

All results in this Part carry the flag **[mult]** indicating they require the multiplication-defect specialization and do not follow from the abstract framework alone.

---

### 3.1  Diagonal vs. Non-Diagonal: The Physical Meaning

**[mult]** For $Q = q(x)$ a multiplication operator and $L$ a local differential operator on $\Omega$, $Q$ is diagonal in position space (it acts by pointwise multiplication) and generically non-diagonal in the spectral coordinates of $L$ (eigenfunctions of $L$ are spatially extended and mix under multiplication by $q$). The commutator $[L, Q]$ is supported on the singular support of $q$ — for $Q = \chi_G$, this is the boundary $\partial G$.

This is the physical content behind the fabrication-as-noncommutativity result: the defect is local in space, the inference coordinates are global (eigenfunctions), and their incompatibility is measured by the commutator $[P_\lambda, Q]$.

---

### 3.2  The Contrast Kernel Formula

**[mult]** For $Q = q(x)$ a bounded measurable function and $R$ a spectral projector with kernel $\Pi_R(x,y)$, for $f \in \mathrm{Ran}\, R$:

**Contrast kernel formula (exact, multiplication-defect realization):**

$$(RMf - Mf)(x) = \int \Pi_R(x,y)\, [q(x) - q(y)]\, f(y)\, dy.$$

The visible finite-resolution artifact is driven by the contrast of the defect profile $q(x)$ across the resolution kernel footprint. Not by missingness, not by locality alone, but by contrast. Hard mask (step function) → jump of 1 across $\partial G$ → strong boundary-layer artifact. Smooth apodization → near-diagonal cancellation in the integrand → weaker artifact with faster off-diagonal decay.

Distance from the hole as seen by finite-band analysis is not Euclidean distance — it is distance as measured by $\Pi_N(x,y)$, the projector kernel geometry, which encodes domain boundaries, reflections, and modal topology.

---

### 3.3  The Three Spatial Objects for the Hard Hole

**[mult]** For $f \in E_\lambda$ with $f \in \mathrm{Ran}\, P_N$ (source mode inside retained band), three objects characterize the hole's effect at every level:

**Object 1 — The Exact Modal Lie:**

$$w_G(f) := (I - P_\lambda) M_G f.$$

The exact off-target modal content. Outside $G$:

$$w_G(x) = \int_G \Pi_\lambda(x,y) f(y)\, dy.$$

For simple spectrum: $w_G(x) = a_n \varphi_n(x)$, $a_n = \int_G |\varphi_n|^2$. Spatially uniform **attenuation** outside $G$ — the multiplicative factor $a_n$ is the same at every point; the pointwise amplitude $|a_n\varphi_n(x)|$ still varies with $x$ through $\varphi_n$. Does not decay with distance from $G$.

**Object 2 — The Commutator Correction (Finite-Band Scatterer):**

$$s_{G,N}(f) := [P_N, M_G] f = -[P_N, \chi_G] f.$$

Outside $G$:

$$s_{G,N}(x) = -\int_G \Pi_N(x,y) f(y)\, dy.$$

Main lobe concentrated near $\partial G$ (width $O(\Lambda^{-1})$), with global oscillatory sidelobes — not a monotone decay. Vanishes as $N \to \infty$ (complete basis). Created entirely by the failure of $P_N$ to commute with $\chi_G$.

**Object 3 — The Finite-Band Phantom Field:**

$$w_{G,N}(f) := (I - P_\lambda) P_N M_G f = P_N w_G(f).$$

Outside $G$:

$$w_{G,N}(x) = \int_G [\Pi_\lambda(x,y) - \Pi_N(x,y)] f(y)\, dy.$$

What a finite-band phantom reconstruction actually displays.

**Exact decomposition (Theorem, established):**

$$w_{G,N} = w_G + s_{G,N}.$$

*Proof:* By hypothesis $f \in E_\lambda \cap \mathrm{Ran}\, P_N$ with $f \neq 0$, so the source eigenvalue satisfies $\lambda \leq \lambda_N$, hence $E_\lambda \subset \mathrm{Ran}\, P_N$. Equivalently $P_N P_\lambda = P_\lambda$, which gives both $[P_N, P_\lambda] = 0$ and $(I - P_N) P_\lambda = 0$.

Step 1 (commutativity gives $w_{G,N} = P_N w_G$):
$w_{G,N} := (I - P_\lambda) P_N M_G f = P_N (I - P_\lambda) M_G f = P_N w_G$ using $[P_N, P_\lambda] = 0$.

Step 2 (split): $P_N w_G = w_G + (P_N - I) w_G$.

Step 3 (evaluate the correction): $(P_N - I)w_G = (P_N - I)(I - P_\lambda) M_G f = (P_N - I) M_G f - (P_N - I) P_\lambda M_G f$. The second term is zero by $(I - P_N) P_\lambda = 0$. The first equals $[P_N, M_G] f = s_{G,N}$: using $f \in \mathrm{Ran}\, P_N$ (so $P_N f = f$), $[P_N, M_G] f = P_N M_G f - M_G P_N f = P_N M_G f - M_G f = (P_N - I) M_G f$. $\square$

### 3.4  Regime Analysis

**Regime I — Complete Basis ($P_N \to I$):**

$s_{G,N} \to 0$. Phantom field is $w_G = a_n \varphi_n(x)$ everywhere outside $G$. Spatially uniform. No near-gap localization.

**Regime II — Intermediate Finite-Band:**

The original observation lives here. $s_{G,N}$ is broad enough to be small relative to $w_G$ at angular positions far from the gap. The visible phantom $w_{G,N}$ appears hole-localized. The "near lobes visible, far lobe missing" pattern arises because near $\partial G$ the main lobe of $s_{G,N}$ adds to $w_G$ and boosts the visible phantom above threshold, while far from $\partial G$ the oscillatory sidelobes of $s_{G,N}$ (with $\sim 1/|x-y|$ envelope — decreasing but not zero) average to a small contribution, so the visible phantom approaches $w_G = a_n\varphi_n$ alone, which may fall below threshold if $a_n$ is small.

**Regime III — High Cutoff:**

As $N \to \infty$, $s_{G,N}$ sharpens to a thin boundary layer at $\partial G$ and weakens elsewhere. Away from $\partial G$, $w_{G,N} \to w_G$. The near-gap localization is strongest at intermediate $N$, not at high $N$. Increasing resolution makes the phantom more spatially uniform. The visible decay is a finite-bandwidth artifact, not an asymptotic truth.

**Regime IV — Point-Defect ($d_G \sqrt{\lambda_N} \ll 1$):**

$$w_{G,N}(x) \approx |G| [\Pi_\lambda(x, x_G) - \Pi_N(x, x_G)] f(x_G).$$

The gap emits the difference of two projector kernels centered at the centroid.

**Regime V — Ambiguity ($\mu_{\max}(C_N) \approx 1$):**

Some retained-band direction concentrates nearly entirely inside $G$. The visible region supports multiple indistinguishable spectral explanations. The gap can sustain a convincing lie.

### 3.5  The 1D Fourier Toy Model

**[mult]** The full framework is exactly solvable on the periodic line. Basis $\{e^{imx}\}$, source $f = e^{ik_0 x}$, hole $G = [-a, a]$:

$$K_{m,k_0} = \frac{\sin((m-k_0)a)}{\pi(m-k_0)}, \qquad a_{k_0} = \frac{a}{\pi}.$$

Outside $G$:

$$w_G(x) = \frac{a}{\pi} e^{ik_0 x},$$

$$s_{G,N}(x) = -\frac{1}{2\pi} \int_{-a}^{a} D_N(x-y) e^{ik_0 y}\, dy,$$

**1D hole calculus (exact):**

$$w_{G,N}(x) = \frac{a}{\pi} e^{ik_0 x} - \frac{1}{2\pi} \int_{-a}^{a} D_N(x-y) e^{ik_0 y}\, dy.$$

Global exact lie plus Dirichlet-kernel correction. Every phenomenon in the framework — near-gap localization, regime-dependent visibility, high-cutoff re-emergence of the uniform lie — is visible in this one line. The general domain replaces $D_N$ with $\Pi_N$.


```python
# 1D regime analysis: w_G, s_GN, and w_GN for three bandwidths
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({'figure.dpi':120,'font.size':10})
BLUE='#1a5fa8'; AMBER='#b85c00'; GREEN='#2a8c3a'; RED='#cc2222'

x = np.linspace(-np.pi, np.pi, 2000)
a_gap = 0.8; k0 = 3

def w_G(x, a, k0):
    return (a/np.pi) * np.cos(k0*x)   # real part of (a/pi)e^{ik0 x}

def s_GN_approx(x, a, k0, N, M=400):
    # Trapezoidal integration of Dirichlet kernel
    y = np.linspace(-a, a, M)
    dy = y[1]-y[0]
    s = np.zeros_like(x)
    for i, xi in enumerate(x):
        diff = xi - y
        # Periodic Dirichlet kernel
        kern = np.where(np.abs(diff % (2*np.pi)) < 1e-10,
                        2*N+1, np.sin((N+0.5)*diff)/np.sin(diff/2))
        s[i] = -np.trapezoid(kern * np.cos(k0*y), y) / (2*np.pi)
    return s

fig, axes = plt.subplots(3, 1, figsize=(13, 11), sharex=True)
for ax, N, col in zip(axes, [4, 15, 60], [BLUE, AMBER, GREEN]):
    wG = w_G(x, a_gap, k0)
    sN = s_GN_approx(x, a_gap, k0, N)
    wGN = wG + sN
    ax.plot(x, wG, '--', color='gray', lw=1.5, alpha=0.7, label='$w_G$ (global lie)')
    ax.plot(x, sN, ':', color=RED, lw=1.5, alpha=0.8, label='$s_{G,N}$ (commutator correction)')
    ax.plot(x, wGN, '-', color=col, lw=2.2, label=f'$w_{{G,N}}$ displayed (N={N})')
    ax.axvspan(-a_gap, a_gap, alpha=0.12, color='silver', label='Gap G')
    ax.axhline(0, color='k', lw=0.5, alpha=0.4)
    ax.legend(loc='upper right', fontsize=8.5, ncol=2)
    regime = {4:'II — intermediate', 15:'II→III — sharpening', 60:'III — high cutoff'}[N]
    ax.set_title(f'Regime {regime}: N={N}', fontsize=10)
axes[-1].set_xlabel('$x$', fontsize=12)
plt.suptitle('Three-Object Decomposition: $w_{G,N} = w_G + s_{G,N}$\n'
             'Near-gap localization strongest at intermediate N, not at high N.', fontsize=11)
plt.tight_layout(pad=1.5)
plt.show()
```

## Part IV — The Three-Part Anatomy of the Hole

The hole problem has three distinct engines, each controlling a different aspect:

### 4.1  Bulk: How Much False Content Exists

Controlled by $B_\lambda = T_{P_\lambda}(\chi_G)$ and its energy law: $\|w_G(f)\|^2 = \langle B_\lambda(I - B_\lambda)f, f\rangle = \sum_j |a_j|^2 \beta_j(1-\beta_j)$.

Design rule (bulk, small-gap limit): minimize $a_n = \int_G |\varphi_n|^2$. Leading-order expansion, centroid at distance $d$ from a nodal line, gap width $a$ in the crossing direction: $a_n \approx |G|\cdot|\nabla\varphi_n(x_\text{line})|^2(d^2 + a^2/12) + O(a^4)$. Scaling depends on gap geometry: slit-like ($|G|\sim L_\| a$) gives $a_n=O(a^3)$ when centered on the line; isotropic 2D ($|G|\sim a^2$) gives $a_n=O(a^4)$. Align the gap cross-section with a nodal line and keep it narrow in the crossing direction.

### 4.2  Edge: Where the False Content Goes in Mode Space

Controlled by the boundary Wronskian routing law (§2.2). Off-diagonal coupling is a boundary pairing on $\partial G$ divided by eigenvalue gap. Near-resonant modes receive more coupling per unit boundary interaction.

Design rule (boundary): minimize $|\varphi_n|_{\partial G}|^2$ — the source amplitude at the gap boundary — to suppress the visible halo. This is distinct from the bulk rule. A gap that clips an antinode at its edge produces strong near-gap artifacts even with small bulk overlap.

### 4.3  Projector: What It Looks Like in Physical Space

The visible artifact is $s_{G,N}(x) = -\int_G \Pi_N(x,y) f(y)\, dy$. The projector kernel $\Pi_N$ acts as a point-spread function on the removed interior field. The artifact has main-lobe concentration near $\partial G$ (width $\sim\lambda_N^{-1/2}$) set by $\Pi_N$, with global oscillatory sidelobes. Main-lobe concentration, not monotone radial decay.

The two design rules (bulk and boundary) are independent. Minimizing bulk overlap controls total phantom energy. Minimizing boundary trace controls the visible near-gap halo. They coincide when the nodal surface passes through the interior of $G$. They diverge when the gap straddles a nodal surface boundary.

## Part V — Recoverability and Multi-View Synthesis

### 5.1  The Three Bases Explicitly

For the retained model class $P_N H$ and a hard hole $Q = \chi_G$, three geometrically distinct bases organize the inverse problem:

- **Exposure basis:** eigenvectors of $T_{P_N}(\chi_G) = C_N$. Sorts retained-band directions by how much of their $L^2$ mass lies inside $G$. Self-adjoint attenuation subclass only.

- **Fabrication basis:** eigenvectors of $\Sigma_{P_N}(\chi_G) = C_N(I - C_N)$. For the hard hole these coincide with the exposure basis ($\Sigma$ is a polynomial in $C_N$). For smooth masks they separate.

- **Ambiguity basis — general $M$:** small-eigenvalue eigenspace of $G_{P_N}(M) = P_N M^*M P_N$. The directions in the retained model class that $M$ nearly erases — that the observation cannot recover.

- **Ambiguity basis — hard hole:** $G_{P_N} = A_N$ (since $Q(I-Q) = 0$), so small eigenvalues of $G_{P_N}$ correspond to large eigenvalues of $C_N$. The ambiguity basis is the near-1 eigenspace of $C_N$. This is the $C_N$-based language used throughout the hole paper.

The three bases are identical for the hard hole (all reduce to the $C_N$ eigenbasis) and generically distinct for soft defects. For the hard hole the ambiguity basis answers: which directions in the retained band concentrate almost entirely inside $G$ and cannot be recovered from the visible region?

---

### 5.2  Recoverability

$A_N = I - C_N$. Condition number: $\kappa_2(A_N) = (1 - \mu_{\min})/(1 - \mu_{\max})$. Diverges as $\mu_{\max} \to 1$.

For noiseless exact recovery: $\mathrm{null}(A_N) = \{0\}$ when $\mu_{\max} < 1$, so $A_N$ is invertible and exact recovery is trivial — direct inversion. Sparsity and $\ell^1$ are regularizers against noise and model mismatch, not identifiability tools.

Noise-robust bound: $\|A_N^*A_N - I\| = \mu_{\max}(2 - \mu_{\max})$. This is the correct RIP-style bound (note: v1 incorrectly wrote $\mu_{\max}(2 + \mu_{\max})$; the correct formula from $\|K^2 - 2K\|$ with eigenvalues $\mu(\mu-2)$ is $\mu_{\max}(2 - \mu_{\max})$, which is bounded by 1 for $\mu_{\max} \in [0,1]$).

---

### 5.3  Multi-View Synthesis

Stack $K$ measurements with different gaps $G_1,\ldots,G_K$ and coupling operators $A_{N,1},\ldots,A_{N,K}$. The stacked normal operator is $\sum_k A_{N,k}^*A_{N,k}$. Stability is controlled by its smallest eigenvalue.

Geometric principle: the ill-conditioned directions of $A_{N,k}$ are those concentrating inside $G_k$ (the ambiguity basis of $C_N^{(k)}$). Stacking helps when these hidden subspaces are mutually transverse. If they align — if the same modal combination hides in every gap — stacking provides no improvement regardless of how many measurements are taken.

For $K$ complementary measurements in the **physical-space partition sense** (non-overlapping physical gaps $\{G_k\}$ with $\sum_k \chi_{G_k} = I$, defining $A_{N,k} = I - P_N\chi_{G_k}P_N$): $\sum_k A_{N,k} = (K-1)I$, giving $\sigma_{\min}(A_{\text{stacked}}) \geq (K-1)/\sqrt{K}$. For $K=2$: $\sigma_{\min} \geq 1/\sqrt{2}$. This is a lower bound, not an equality — the stacked system can be strictly better when concentration eigenvalues are not at $1/2$. (A different bound, $\sigma_{\min}\geq\sqrt{K-1}$, applies to the **rank-1-complement partition** setup where each $A_k = I - e_k e_k^*$ with orthonormal $\{e_k\}$ in mode space; see `hole_problem.ipynb` §8.3. The two setups are both called "complementary" but differ: for $K=2$, physical-space gives $1/\sqrt 2$, rank-1-complement gives $1$.)


```python
# Recoverability and multi-view stacking
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({'figure.dpi':120,'font.size':10})
BLUE='#1a5fa8'; AMBER='#b85c00'; GREEN='#2a8c3a'; RED='#cc2222'

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Bernoulli variance
beta = np.linspace(0, 1, 300)
axes[0].plot(beta, beta*(1-beta), color=BLUE, lw=2.5)
axes[0].axvline(0.5, color=RED, lw=1.5, ls='--', label='Max at beta=1/2')
axes[0].fill_between(beta, beta*(1-beta), alpha=0.15, color=BLUE)
axes[0].set_xlabel('beta = a_n (fraction inside G)')
axes[0].set_ylabel('Fabrication energy Sigma_P = beta(1-beta)')
axes[0].set_title('Bernoulli Variance\nPartial occlusion is worst case')
axes[0].legend(fontsize=9)

# RIP bound -- corrected vs wrong
mu = np.linspace(0, 1, 300)
axes[1].plot(mu, mu*(2-mu), color=GREEN, lw=2.5, label='Correct: mu(2-mu)')
axes[1].plot(mu, mu*(2+mu), color=RED, lw=2, ls='--', alpha=0.7, label='v1 error: mu(2+mu)')
axes[1].axhline(1, color='gray', lw=1, ls=':', label='RIP bound = 1')
axes[1].set_xlabel('mu_max(C_N)')
axes[1].set_ylabel('||A_N*A_N - I||')
axes[1].set_title('Noise-Robust Bound (RIP-style)\nv1 sign error corrected')
axes[1].legend(fontsize=9)
axes[1].set_ylim(0, 3)

# Multi-view stacking
K_vals = np.arange(2, 12)
sigma_lb = (K_vals - 1) / np.sqrt(K_vals)
axes[2].plot(K_vals, sigma_lb, 'o-', color=AMBER, lw=2, markersize=7)
axes[2].axhline(1/np.sqrt(2), color=BLUE, lw=1.5, ls='--', alpha=0.8,
               label=f'K=2: 1/sqrt(2) = {1/np.sqrt(2):.3f}')
axes[2].set_xlabel('Number of complementary measurements K')
axes[2].set_ylabel('sigma_min(A_stacked) lower bound')
axes[2].set_title('Multi-View Stacking: (K-1)/sqrt(K)\n(lower bound, not equality)')
axes[2].legend(fontsize=9)

plt.suptitle('Recoverability Geometry', fontsize=11, y=1.02)
plt.tight_layout(pad=1.5)
plt.show()
```

## Part VI — Downstream Observable Distortion

### 6.1  The Pullback Distortion Calculus

The pipeline computes $\mathcal{O}(A_P f)$ when the intended quantity is $\mathcal{O}(f)$. The defect-induced distortion is:

$$\delta_{M,P} \mathcal{O} := \mathcal{O} \circ A_P(M) - \mathcal{O}.$$

For Fréchet-differentiable $\mathcal{O}$:

$$\mathcal{O}(A_P f) - \mathcal{O}(f) = D\mathcal{O}_f[(A_P - I)f] + \tfrac{1}{2} D^2\mathcal{O}_f[(A_P-I)f, (A_P-I)f] + \ldots$$

Linear observables see first-order contamination. Quadratic diagnostics (coherence, interaction matrices, bispectra) see bilinear spreading. The mask acts before interpretation, so every downstream functional is composed with the wrong forward map.

### 6.2  Quadratic Observables: The Interaction Matrix

For interaction matrix $\mathcal{O}(c) = cc^*$: $\hat{I} = \hat{c}\hat{c}^* = A_P(cc^*)A_P^*$. A rank-$r$ true state produces a generically full-rank observed interaction. Every mode-pair $(m,n)$ receives a false cross-term even when the true field has no content at $m$ or $n$. The false interaction is deterministic and predicted exactly by the coupling matrix $K$.

### 6.3  Temporal Inheritance Detection

For time-evolving $f(x,t)$ with static mask: $\hat{c}_m(t) = A_{mn_0} c(t)$. Every phantom coefficient inherits the source mode's temporal dynamics, not the dynamics expected at eigenvalue $\lambda_m$. Three detection tests:

- **Dispersion test:** temporal FFT of $\hat{c}_m(t)$ peaks at $\omega_{n_0}$ not $\omega_m$. Single FFT per coefficient.

- **Coherence test:** $|\gamma^2(\hat{c}_m, \hat{c}_{n_0})| = 1$ for phantoms. Geometry-independent — works without knowing gap geometry.

- **K-amplitude test:** $|\hat{c}_m|/|\hat{c}_{n_0}| = |K_{mn_0}|/|1 - K_{n_0 n_0}|$. Requires known gap geometry. Quantitative acceptance criterion.

All three tests are exact for a static mask. For time-varying masks, temporal inheritance breaks and phantoms acquire sidebands — dispersion and coherence tests can produce false negatives.

## Part VII — The Multipole Expansion

### 7.1  Beyond the Monopole

**[mult]** The rank-one (monopole) approximation for a small gap centered at $x_G$:

$$K_{mn} \approx |G| \varphi_n(x_G) \bar{\varphi}_m(x_G).$$

The full Taylor expansion of $\varphi_n(y)\bar{\varphi}_m(y)$ about $x_G$, integrated over $G$:

**Multipole expansion of the coupling matrix:**

$$K_{mn} = |G| F_{mn}(x_G) + M_j \partial_j F_{mn}(x_G) + \tfrac{1}{2} Q_{jk} \partial_{jk} F_{mn}(x_G) + \ldots,$$
$$F_{mn}(x) = \varphi_n(x)\bar{\varphi}_m(x),$$
$$M_j = \int_G (y-x_G)_j\, dy \quad \text{(first moments, vanish if } G \text{ centrosymmetric)},$$
$$Q_{jk} = \int_G (y-x_G)_j(y-x_G)_k\, dy \quad \text{(second moments, encode shape/orientation)}.$$

*Correction from v1: v1 stated the Taylor expansion is "valid when $x_G$ is not on a nodal set." This is wrong. The product $\varphi_n(y)\bar{\varphi}_m(y)$ is smooth on the interior of $\Omega$ regardless of whether $x_G$ is a nodal point — eigenfunctions of elliptic operators are smooth on the interior. What happens at a nodal point is that the monopole coefficient $F_{mn}(x_G) = \varphi_n(x_G)\bar{\varphi}_m(x_G) = 0$, so the monopole term vanishes and the dipole becomes the leading contribution. The expansion exists everywhere; the monopole term may vanish.*

Two gaps with the same area and centroid are indistinguishable at monopole order. They first differ at quadrupole order, where gap shape and orientation enter. This is a concrete, computable distinction.

### 7.2  Degenerate Eigenspaces: Visibility Selection

**[mult]** For a $d$-dimensional eigenspace $E_\lambda$ and a small gap at $x_G$, the leading-order exposure operator is:

$$B_\lambda \approx |G| v_\lambda v_\lambda^*, \qquad v_\lambda = (u_1(x_G), \ldots, u_d(x_G)).$$

*Correction from v1: the visibility-rank theorem should read: $B_\lambda$ has rank at most one to leading order, and rank exactly one when $v_\lambda \neq 0$. If all eigenspace basis functions vanish at $x_G$ (i.e., $x_G$ lies on the nodal set of the entire eigenspace), then the monopole term is zero, the rank-one approximation gives $B_\lambda \approx 0$, and the first nontrivial contribution comes from dipole terms. The star-graph example is the clean case where one antisymmetric direction has zero amplitude on the gapped edge ($\beta_2 = 0$ exactly) and the other does not ($\beta_1 = (2/3)\rho > 0$).*

The angular selectivity $\beta_+ - \beta_- = 2\rho|\sin(n\Delta\theta)|/n$ for the disk measures the gap's ability to distinguish degenerate directions. The accidental zeros at $n\Delta\theta = k\pi$ (gap subtends an integer number of half-periods) are the values where the gap exposes both directions equally and has no angular discriminating power.

## Part VIII — Three Engineering Deliverables

The framework produces three concrete engineering tools that are not available in current practice. These are the outputs that matter if the mathematics is to be useful.

### 8.1  The Gap Susceptibility Map

Before any data collection: compute $|K_{mn}|$ for your gap geometry and eigenfunctions. This gives a pre-measurement calibration table — if mode $n$ is excited, which modal indices show false peaks, and at what amplitude. Any spectral analysis pipeline operating on gapped data should compute this and flag susceptible bins. The computation requires the eigenfunctions (from FE model or analytic formula) and the gap geometry. The output is deterministic and geometry-dependent, not statistical.

Limitation: $K$ is computed from model eigenfunctions. If the FE model is wrong, $K$ is wrong. The framework is exact given the eigenfunctions; it inherits all model uncertainty.

### 8.2  The Dual Design Rule

Standard practice: minimize $\int_G |\varphi_n|^2$ (bulk rule). The framework adds: also minimize $|\varphi_n|_{\partial G}|^2$ (boundary trace rule). These are different objectives and diverge when the gap straddles a nodal surface. The boundary trace controls the visible near-gap artifact through the projector kernel; the bulk controls the total phantom energy budget. An instrument designed only on the bulk rule can produce strong visible artifacts from gaps that clip antinodes at their boundaries.

### 8.3  The Forward Check: $\hat{I} = AIA^*$

Given gap geometry and estimated source spectrum: compute the predicted contaminated interaction matrix $\hat{I}_{\text{predicted}} = A I_{\text{estimated}} A^*$. Compare to the measured interaction matrix $\hat{I}_{\text{measured}}$. The residual $\hat{I}_{\text{measured}} - \hat{I}_{\text{predicted}}$, if larger than predicted by $K$, signals physics beyond gap contamination. This turns a qualitative "beware of gap artifacts" warning into a quantitative subtraction step and a residual diagnostic.

Limitation: this requires both the gap geometry (to compute $A$) and an estimate of the source spectrum (to compute $I_{\text{estimated}}$). In practice both have uncertainty. The forward check is most useful in controlled settings where the gap is well-characterized.

## Part IX — Theorem Program

**[E]** = algebraic identity, established.  **[K]** = follows from known results in cited literature.  **[O]** = open theorem candidate.

Theorem status is labeled honestly. [E] = algebraic identity, verifiable by expansion. [K] = follows from known results in the cited literature with no significant gap. [O] = open theorem candidate with identified ingredients but missing formal bridge.

**[E]  T1 — Compression Variance Identity**

$$\Sigma_P(M) = H_P(M)^*H_P(M) = G_P(M) - A_P(M)^*A_P(M).$$

Algebraic identity for general bounded $M$ and orthogonal projection $P$. In the self-adjoint attenuation case $M = I-Q$: $\Sigma_P(Q) = PQ^2P - (PQP)^2$. Proved in Part I.

---

**[E]  T2 — Forward-Measurement Geometry Identity**

$$G_P(M) = A_P(M)^*A_P(M) + \Sigma_P(M).$$

Algebraic identity. For $Q^2 = Q$ (hard hole): $Q(I-Q) = 0$, so $G_P = A_P$ exactly. For general $Q$: $G_P = A_P$ iff $PQ(I-Q)P = 0$, not iff $\Sigma_P(Q) = 0$ (these differ for $Q = \alpha I$ as the counterexample shows). Proved in Part I.

---

**[E]  T3 — Decomposition Theorem**

$$w_{G,N} = w_G + s_{G,N}.$$

Exact for $f \in E_\lambda \cap \mathrm{Ran}\, P_N$. **[mult]** Proved in Part III §3.3.

---

**[E]  T4 — Commutator Routing Law**

$$(\mu - \lambda)^k Q_{\mu\lambda} = P_\mu \, \mathrm{ad}_L^k(Q) \, P_\lambda.$$

Algebraic identity from the spectral theorem. Decay estimate $\|Q_{\mu\lambda}\| \leq \|\mathrm{ad}_L^k(Q)\| / |\mu-\lambda|^k$ follows immediately when $\mathrm{ad}_L^k(Q)$ is bounded. Domain qualification required for unbounded $L$ — see §2.1 note.

---

**[E]  T5 — Contrast Kernel Formula**

$$(RMf - Mf)(x) = \int \Pi_R(x,y)\, [q(x) - q(y)]\, f(y)\, dy \qquad (f \in \mathrm{Ran}\, R).$$

**[mult]** Exact for multiplication defects $Q = q(x)$. Proved in Part III §3.2.

---

**[O]  T6 — Boundary-Layer Localization**

For $Q = \chi_G$, the commutator correction $s_{G,N}$ concentrates in a tubular neighborhood of $\partial G$ of width $O(\lambda_N^{-1/2})$, with amplitude controlled by $|f|_{\partial G}|$.

Status: open. Ingredients: decay of $\Pi_N(x,y)$ off the diagonal (requires Hörmander's local Weyl law for the specific domain/operator), combined with the fact that $q(x)-q(y) = \pm 1$ only near $\partial G$. The qualitative concentration is plausible and consistent with all examples. The quantitative tube width and profile need the spectral projector estimates applied explicitly to this defect class. The genesis document correctly identified this as unproved; v1 of this document incorrectly upgraded it to [K].

---

**[O]  T7 — Basis Bifurcation**

For generic positive contractions $Q$ with $Q^2 \neq Q$, the exposure basis (eigenvectors of $T_P(Q)$) and fabrication basis (eigenvectors of $\Sigma_P(Q)$) are generically distinct. The set of $Q$ where they coincide has codimension $\geq 1$ in the space of positive contractions.

Status: open. The coincidence is forced by $Q^2 = Q$ (projection defects) and by $Q$ proportional to $I$ (scalar multiples). What is the complete characterization? The codimension claim requires a genericity argument — plausible but not proved. Active research direction.

---

**[K]  T8 — Multipole Expansion with Remainder**

$$K_{mn} = \sum_{|\alpha|\leq k} m_\alpha \partial^\alpha(\varphi_n \bar{\varphi}_m)(x_G) + O(|G|\, d_G^{k+1} \sup|\partial^{k+1}(\varphi_n\bar{\varphi}_m)|).$$

**[mult]** Follows from Taylor's theorem applied to $F_{mn}(y) = \varphi_n(y)\bar{\varphi}_m(y)$ on $G$ about $x_G$. Eigenfunctions of elliptic operators are smooth on the interior, so $F_{mn}$ is smooth and the expansion exists everywhere. The monopole term vanishes when $x_G$ is a nodal point of $F_{mn}$; higher moments then lead. Remainder bound is standard Taylor remainder. Full derivation with explicit constants for specific geometries (rectangle, disk) is routine but not yet written out.

---

**[K]  T9 — Visibility-Rank Theorem for Degeneracies**

**[mult]** For a $d$-dimensional eigenspace $E_\lambda$ and a gap $G$ with $d_G\sqrt{\lambda_N} \ll 1$, $B_\lambda = T_{P_\lambda}(\chi_G)$ has rank at most one at monopole order, and rank exactly one when $v_\lambda = (u_1(x_G),\ldots,u_d(x_G)) \neq 0$. The first nontrivial correction is the dipole term; for centrosymmetric gaps the first correction is the quadrupole term. Follows from the multipole expansion T8.

---

**[O]  T10 — Multi-View Transversality Lower Bound**

$\sigma_{\min}([A_1;\ldots;A_K])^2 \geq f(\theta_{\min})$, where $\theta_{\min}$ is the minimum principal angle between the ambiguity bases (near-1 eigenspaces of $C_N^{(k)}$), and $f : [0, \pi/2] \to [0,\infty)$ with $f(0) = 0$ and $f(\pi/2) > 0$.

Status: open. The geometric picture — transversality of ambiguity subspaces controls stacked conditioning — is correct and the direction of the bound is clear. The specific form of $f(\theta)$ requires working out how principal angles between concentration eigenspaces translate to singular values of the stacked normal operator $\sum_k A_k^*A_k$. This is the hard part. Numerical evidence from the companion script supports the principle but does not establish the bound.
